**1. Import Required Libraries & Setup**

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set display options
pd.set_option('display.max_columns', None)
sns.set_style("whitegrid")

# Load environment variable if needed
from dotenv import load_dotenv
load_dotenv()

# Use project root relative path
DATA_PATH = os.getenv("RAW_DATA_PATH", "../data/raw/AmesHousing.csv")

# Load dataset
df = pd.read_csv(DATA_PATH)

print(f"Dataset loaded successfully with shape: {df.shape}")
df.head()
```

**2. Basic Dataset Information**

```python
# General info about dataset
df.info()

# Summary statistics
df.describe().T.head(10)
```

**3. Missing Values & Null Percentages**

```python
# Calculate missing value percentage
missing = df.isnull().mean() * 100
missing = missing[missing > 0].sort_values(ascending=False)

print(f"Total columns with missing values: {len(missing)}")

plt.figure(figsize=(10,5))
sns.barplot(x=missing.values, y=missing.index, color="skyblue")
plt.title("Columns with Missing Values (%)")
plt.xlabel("% Missing")
plt.ylabel("Feature")
plt.show()

missing.head(10)
```

4. Drop Irrelevant / Useless Columns

```python
# Drop columns that have no predictive value or are identifiers
drop_cols = ['Order', 'PID']

# Also drop columns with >40% missing data
too_missing = missing[missing > 40].index.tolist()
drop_cols.extend(too_missing)

print("Dropping columns:", drop_cols)
df.drop(columns=drop_cols, inplace=True, errors='ignore')
print(f"Remaining columns: {df.shape[1]}")
df.head()
```

5. Handle Missing Values

```python
# Separate numeric and categorical columns
num_cols = df.select_dtypes(include=np.number).columns
cat_cols = df.select_dtypes(exclude=np.number).columns

# Fill numeric NaNs with median
df[num_cols] = df[num_cols].apply(lambda x: x.fillna(x.median()))

# Fill categorical NaNs with mode or 'None'
for col in cat_cols:
    df[col].fillna(df[col].mode()[0] if not df[col].mode().empty else "None", inplace=True)

print("Missing values handled successfully!")
```

6. Check Duplicates and Basic Sanity

```python
# Separate numeric and categorical columns
num_cols = df.select_dtypes(include=np.number).columns
cat_cols = df.select_dtypes(exclude=np.number).columns

# Fill numeric NaNs with median
df[num_cols] = df[num_cols].apply(lambda x: x.fillna(x.median()))

# Fill categorical NaNs with mode or 'None'
for col in cat_cols:
    df[col].fillna(df[col].mode()[0] if not df[col].mode().empty else "None", inplace=True)

print("✅ Missing values handled successfully!")
```

7. Quick Statistical Snapshot

```python
# Show key statistics
display(df.describe().T.head(15))

# Count of categorical variables
for col in cat_cols[:5]:
    print(f"\n{col} unique values: {df[col].nunique()}")
    print(df[col].value_counts().head())
```

8. Save Cleaned Version (Initial Stage)

```python
# Save the cleaned version (stage 1)
os.makedirs("../data/processed", exist_ok=True)
df.to_csv("../data/processed/AmesHousing_stage1.csv", index=False)

print("Cleaned dataset (Stage 1) saved to: data/processed/AmesHousing_stage1.csv")
```
